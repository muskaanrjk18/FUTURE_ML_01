```python
dataset_path = os.path.join(data_path, "sales_data.csv")

df = pd.read_csv(dataset_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nDataset Summary:")
df.describe(include="all")
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check missing values
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"])

print(df[["Order_Date", "Ship_Date"]].dtypes)
```

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day

df[["Order_Date", "Year", "Month", "Day"]].head()
```

```python
monthly_sales = df.groupby(
    df["Order_Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales = monthly_sales.sort_index()

monthly_sales.head(10)
```

```python
monthly_df = monthly_sales.reset_index()

monthly_df.columns = ["Month", "Sales"]

monthly_df["Month"] = monthly_df["Month"].dt.to_timestamp()

monthly_df["Time"] = np.arange(len(monthly_df))

display(monthly_df.head())
print("Total Months:", len(monthly_df))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_df["Month"], monthly_df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
future_time = pd.DataFrame({
    "Time": np.arange(
        len(monthly_df),
        len(monthly_df) + 3
    )
})

future_predictions = model.predict(future_time)

future_dates = pd.date_range(
    start=monthly_df["Month"].max() + pd.DateOffset(months=1),
    periods=3,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Month": future_dates,
    "Forecasted_Sales": future_predictions
})

display(forecast_df)
```
